# 1. train test split 

```python
from sklearn.model_selection import train_test_split
x_train, x_text, y_train,y_test = train_test_split(df.drop('Purchased', axis=1),
                                                   df['Purchased'],
                                                   test_size = 0.3,
                                                   random_state = 0)

x_train.shape, x_text.shape
```

# 2. srtandardScaler 

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# fit the scaler to the train set it will learn the parameters

scaler.fit(x_train)

#transform train and test sets 
x_train_scaled = scaler.transform(x_train)
x_text_scaled = scaler.transform(x_text)
```

```python
scaler.mean_
```

```python
x_train
```

```python
x_train_scaled
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns=x_train.columns)
x_text_scaled = pd.DataFrame(x_text_scaled, columns=x_text.columns)
```

# 3. Effect of Scaling

```python
fig, (ax1 ,ax2) = plt.subplots (ncols=2, figsize=(12,5))

ax1.scatter(x_train['Age'], x_train['EstimatedSalary'])
ax1.set_title("before scaling")

ax2.scatter(x_train_scaled['Age'], x_train_scaled['EstimatedSalary'], color='red')
ax2.set_title("arter scaling ")

plt.show()
```

```python
fig, (ax1 ,ax2) = plt.subplots (ncols=2, figsize=(12,5))

#before scaling 
ax1.set_title("before scaling")
sns.kdeplot(x_train['Age'], ax = ax1)
sns.kdeplot(x_train['EstimatedSalary'], ax = ax1)

#after scaling 
ax2.set_title("before standard scaling")
sns.kdeplot(x_train_scaled['Age'], ax = ax2)
sns.kdeplot(x_train_scaled['EstimatedSalary'], ax = ax2, color='red')

plt.show()
```

# 3. Comparison of Distributions

```python
fig, (ax1, ax2 ) = plt.subplots(ncols=2, figsize=(12,5))

# before scaling
ax1.set_title("age distribution before scaling ")
sns.kdeplot(x_train['Age'], ax= ax1)

# after scaling 
ax2.set_title("age distributon after standard scaling ")
sns.kdeplot(x_train_scaled['Age'], ax=ax2)

plt.show()
```

```python
fig, (ax1, ax2 ) = plt.subplots(ncols=2, figsize=(12,5))

# before scaling
ax1.set_title("salary distribution before scaling ")
sns.kdeplot(x_train['EstimatedSalary'], ax= ax1)

# after scaling 
ax2.set_title("salary distributon after standard scaling ")
sns.kdeplot(x_train_scaled['EstimatedSalary'], ax=ax2)

plt.show()
```

# why scaling is important ? 

```python
from sklearn.linear_model import LogisticRegression
```

```python
il = LogisticRegression()
il_scaled = LogisticRegression()
```

```python
il.fit(x_train, y_train)
il_scaled.fit(x_train_scaled, y_train)
```

```python
y_pred = il.predict(x_text)
y_pred_scaled = il_scaled.predict(x_text_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("actual", accuracy_score(y_test, y_pred))
print("scaled", accuracy_score(y_test, y_pred_scaled))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt_scaled = DecisionTreeClassifier()
```

```python
dt.fit(x_train, y_train)
dt_scaled.fit(x_train_scaled,y_train)
```

```python
y_pred = dt.predict(x_text)
y_pred_scaled = dt_scaled.predict(x_text_scaled)
```

```python
print("actual", accuracy_score(y_test, y_pred))
print("scaled", accuracy_score(y_test, y_pred_scaled))
```
