```python
from sklearn.datasets import make_classification
import numpy as np
x,y = make_classification(n_samples=100,n_features=2, n_informative=1, n_redundant=0, n_classes=2, n_clusters_per_class=1, random_state=41, hypercube=False, class_sep=20)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(x[:,0], x[:,1], c=y, cmap='winter', s=100)
```

```python
from sklearn.linear_model import LogisticRegression
log = LogisticRegression(penalty='None', solver='sag')
log = LogisticRegression(penalty=None, solver='sag')
log.fit(x, y)
```

```python
print(log.coef_)
print(log.intercept_)
```

```python
m1 = -(log.coef_[0][0]/log.coef_[0][1])
b1 = -(log.intercept_/log.coef_[0][1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m1*x_input + b1
```

```python
def gd(x,y):
    x= np.insert(x, 0, 1, axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.5

    for i in range(2500):
        y_hat = sigmoid(np.dot(x, weights))
        weights = weights +lr*(np.dot((y-y_hat), x)/x.shape[0])

    return weights
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
weights = gd(x,y)
intercept_ = weights[0]
coef_ = weights[1:]
# print(coef_)
# print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3, 3, 100)
y_input1 = m*x_input1 + b
```

```python
plt.Figure(figsize=(10,6))
plt.plot(x_input1, y_input1, color='red', label='Gradient Descent', linewidth=2)
plt.plot(x_input, y_input, color='blue', label='Logistic Regression', linewidth=2)
plt.scatter(x[:,0], x[:,1], c=y, cmap='winter', s=100) 
plt.ylim(-3, 2)
```
