```python
df = pd.read_csv('wine_data.csv', header=None,usecols=[0,1,2])
df.columns = ['Class', 'Alcohol', 'Malic acid']
df.head()
```

```python
df
```

```python
# sns.kdeplot(df['Alcohol'])/
df['Alcohol'].dtype
```

```python
df['Alcohol'] = pd.to_numeric(df['Alcohol'], errors='coerce')

sns.kdeplot(x=df['Alcohol'])
```

```python
df['Malic acid'] = pd.to_numeric(df['Malic acid'], errors='coerce')

sns.kdeplot(x=df['Malic acid'])
```

```python
color_dict = {1: 'red', 2: 'green', 3: 'blue'}
sns.scatterplot(x=df['Alcohol'], y=df['Malic acid'], hue=df['Class'], palette=color_dict)
```

```python
df.head()
```

```python
print(df.columns)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.drop('Class', axis=1),
                                                    df['Class'],
                                                    test_size=0.3,
                                                    random_state= 0)
x_train.shape, x_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
# fit the scaler to the train set it well learn the parameters 
scaler.fit(x_train)
# transform the train and test set
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns=x_test.columns)
```

```python
np.round(x_train.describe(), 1)
```

```python
np.round(x_train_scaled.describe(), 1)
```

```python

fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,5))

ax1.scatter(x_train['Alcohol'],x_train['Malic acid'],c =y_train)
ax1.set_title("before scaling")

ax2.scatter(x_train_scaled['Alcohol'],x_train_scaled['Malic acid'], c=y_train)
ax2.set_title("after scaling")

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize= (12,5))

#Before scaling
ax1.set_title("before scaling")
sns.kdeplot(x_train['Alcohol'], ax=ax1)
sns.kdeplot(x_train['Malic acid'], ax=ax1)

#After scaling
ax2.set_title("after scaling")
sns.kdeplot(x_train_scaled['Alcohol'], ax=ax2)
sns.kdeplot(x_train_scaled['Malic acid'], ax=ax2)

plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(ncols=2, figsize=(12,5))

# before scaling
ax1.set_title("alcohol distribution before scaling ")
sns.kdeplot(x_train['Alcohol'], ax=ax1)

# after scaling

ax2.set_title("alcohol distribution afret scaling")
sns.kdeplot(x_train_scaled['Alcohol'], ax=ax2)
```

```python
fig, (ax1,ax2) = plt.subplots(ncols=2, figsize=(12,5))

# before scaling
ax1.set_title("Malic acid distribution before scaling ")
sns.kdeplot(x_train['Malic acid'], ax=ax1)

# after scaling

ax2.set_title("Malic acid distribution afret scaling")
sns.kdeplot(x_train_scaled['Malic acid'], ax=ax2)
```
