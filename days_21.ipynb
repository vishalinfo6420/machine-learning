```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

1.  Sacttreplot 

### (numeric to numeric )

```python

# sns.scatterplot(x=tips['total_bill'], y=tips['tip'])
sns.scatterplot(data=tips, x='total_bill', y='tip', hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

```python
titanic = pd.read_csv("titanic.csv")
# 
titanic.head()
```

1. baroplot (nummeric to categorical data)

```python
sns.barplot(x= titanic['Pclass'], y = titanic['Age'], hue=titanic['Sex'], data=titanic)
```

### 1. boxplot (numeric data to categorical data) 

```python
sns.boxplot(x=titanic['Pclass'], y=titanic['Age'], hue=titanic['Sex'], data=titanic)
```

### 1. distplot (numeric data to categorical data) 

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
```

### 1.  heatmap( categorical data to categorical data) 

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
titanic.groupby('Pclass')['Survived'].mean() * 100
```

## 1. clustermap ( categorical data to  categorical data)

```python
# sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived'],figsize=(8, 4)))
sns.clustermap(
    pd.crosstab(titanic['Parch'], titanic['Survived']),
    figsize=(4, 4)
)
```
